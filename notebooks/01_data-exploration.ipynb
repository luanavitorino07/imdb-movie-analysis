## Verificação da qualidade dos dados

Nessa etapa, procuramos por possívels problemas na base de dados que podem ser tratados na etapa de limpeza.

### Valores nulos

```python
null_count = raw_imdb_df.isna().sum()

null_summary = pd.DataFrame({
    "Quantidade de Nulos": null_count,
    "Porcentagem": ( null_count / len(raw_imdb_df)) * 100 
})

null_summary.sort_values(
    by="Porcentagem",
    ascending=False
)
```

As colunas `Certificate`, `Meta_score` e `Gross` apresentam entre 10% e 20% de valores nulos entre os preenchimentos. É necessária uma análise mais detalhada dessas entradas com valores nulos, a fim de entender o quanto prejudicam a análise e avaliar a possibilidade de preenchê-las.

### Valores duplicados

```python
duplicate_count = raw_imdb_df.duplicated().sum()

pd.Series(
    {"Registros duplicados": duplicate_count}
)
```

Não existem valores duplicados na base de dados.

### Distribuição das variáveis categóricas

```python
raw_imdb_df["Certificate"].value_counts()
```

```python
raw_imdb_df["Genre"].value_counts()
```

A partir dessa distribuição simples das variáveis categóricas, podemos fazer duas observações:
 1) A coluna `Certificate`, referente à classificação indicativa dos filmes, mistura padrões de classificação de diversos países. Esse detalhe deve ser corrigido na etapa de limpeza, com a adoção de apenas um padrão para todos os filmes.
 2) A coluna `Genre` salva os gêneros dos filmes como uma `string`, com cada gêreno separado por vírgula. Isso pode dificultar análises relacionadas à categoria, sendo necessário transformar o formato da coluna em uma lista de gêneros durante a etapa de limpeza para uma análise mais precisa.

### Verificação de possíveis valores inconsistentes
